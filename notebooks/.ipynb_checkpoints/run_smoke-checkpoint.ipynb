```python
import os
import sys
from pathlib import Path

def find_file(name: str) -> Path:
    roots = [Path("/kaggle/input"), Path("/kaggle/working"), Path(".")]
    for root in roots:
        if not root.exists():
            continue
        direct = root / name
        if direct.is_file():
            return direct
        # one or two levels deep
        for p in root.rglob(name):
            if p.is_file():
                return p
    raise FileNotFoundError(name)

code_py = find_file("run_smoke.py")
code_root = code_py.parent
sys.path.insert(0, str(code_root))
print("code:", code_root)

train_path = find_file("train.jsonl")
val_path = find_file("val.jsonl")
print("train:", train_path)
print("val:", val_path)
```

```python
from run_smoke import run_smoke
import argparse

# Short smoke: ~200 steps × batch1 × seq512 × accum8 ≈ ~0.8M tokens order-of-magnitude
# Raise --max-steps toward 1–5M tokens after the loop is stable.
ns = argparse.Namespace(
    model_name=None,
    train_data=str(train_path),
    val_data=str(val_path),
    output_dir="/kaggle/working/checkpoints",
    seq_length=512,
    batch_size=1,
    max_steps=200,
    max_eval_batches=20,
    max_train_texts=None,
    max_val_texts=None,
    skip_train=False,
    skip_shock=False,
    no_bf16=False,
    no_8bit_adam=False,
    quant_warmup_steps=100,
    seed=42,
    device_map="auto",
)
results = run_smoke(ns)
print({k: results[k] for k in results if k.startswith("ppl") or k == "loss_finite"})
```
